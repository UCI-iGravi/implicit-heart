```python
# Standard libraries
import glob
import os
import re
import csv

# Third-party libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pydicom import dcmread
import cv2
import skimage.morphology
from skimage.morphology import disk
from skimage.filters import median

# Local libraries
from modules.scd import data_utils, file_utils, plot_utils

# My libraries
from modules.implicit import ImplicitFunction
import modules.utils as utils
```

```python
# Define paths
scd_csv_path = "data/scd/scd_patientdata.csv"
scd_contours_path = "data/scd/SCD_ManualContours/"
scd_image_path = "data/scd/SCD_IMAGES/"

# User input
#patient = "SCD0004101"
```

Load and setup the required files

```python
# Load the patient csv information
scd_csv = pd.read_csv(scd_csv_path)

patient_ids = list(scd_csv['PatientID'])

output_path = 'output/scd/' 
try:
    os.mkdir(output_path)
except OSError as error:
    print(output_path + ' already exists.')

for patient in patient_ids:
    # Get patient's ID from the image path and get the corresponding contours folder
    contour_folder = scd_contours_path + file_utils.get_patient_id(scd_csv, patient) + '/contours-manual/IRCCI-expert/'

    # Get every contour file for the patient
    icontour_dict, ocontour_dict = file_utils.load_contours(contour_folder, (256, 256), 
                                                            verbose=True)

    icontour_keys = sorted(icontour_dict.keys())
    ocontour_keys = sorted(ocontour_dict.keys())
    # Print loaded file information
    print('\nPatient:', patient)
    print('Contours folder:', contour_folder)
    print()
    print('icontour indices:', list(icontour_keys))
    print('ocontour indices:', list(ocontour_keys))

    out_volume = []
    original_volume = []
    for i in range(len(icontour_keys) - 1):
        curr_key = icontour_keys[i]
        next_key = icontour_keys[i+1]
        current_icontour = data_utils.get_contour(icontour_dict, curr_key, verbose=False)
        next_icontour = data_utils.get_contour(icontour_dict, next_key, verbose=False)
        original_volume.append(current_icontour)
        #current_ocontour = data_utils.get_contour(ocontour_dict, curr_key, verbose=False)
        #next_ocontour = data_utils.get_contour(ocontour_dict, next_key, verbo)
        imp_in = np.array([current_icontour, next_icontour])
        imp = ImplicitFunction(imp_in, tmax=1)
        out = imp.interpolate(0)
        out = median(out, disk(3))
        out = skimage.morphology.remove_small_objects(out.astype(bool))
        out = skimage.morphology.remove_small_holes(out.astype(bool))
        out_volume.append(out)
        #plt.imshow(current_icontour)
        #plt.imshow(current_ocontour)
        #plt.show()
    # Last slice
    original_volume.append(next_icontour)
    out = imp.interpolate(1)
    out_volume.append(out)
    out_volume = np.array(out_volume)
    original_volume = np.array(original_volume)
    np.save(output_path + '/' + patient + '_interp.npy', out_volume)
    np.save(output_path + '/' + patient + '_original.npy', original_volume)
```
